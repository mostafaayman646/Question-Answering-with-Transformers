title (one title for row)

Paragraphs (one par for row)
- qas (many qas for par)
  - question (many questions for qas)
  - id (id for each quest)
  - answers (one answer for each ques)
    - text (one text for answer)
    - answer_start (one index)
  - is_impossible (True or false for each answer)
- context (single context for each qas)

```python
sum(qa['is_impossible'] for p in df.iloc[0, 1]['paragraphs'] for qa in p['qas'])
```

```python
missing_context = 0

for i, p in enumerate(df.iloc[0, 1]['paragraphs']):
    # Check if 'context' is missing, None, or an empty string
    if 'context' not in p or not p.get('context') or p['context'].strip() == "":
        print(f"Paragraph {i} is missing context! It has {len(p.get('qas', []))} questions.")
        missing_context += 1

print(f"Total paragraphs missing context: {missing_context}")
```

```python
# Loop through paragraphs and questions
for paragraph in df.iloc[300, 1]['paragraphs']:
    for qa in paragraph['qas']:
        if qa.get('is_impossible') == True:
            print("CONTEXT:")
            print(paragraph['context'])
            print("\nQUESTION DATA:")
            print(qa)
            
            break  # Stop at the inner loop once we find one
    else:
        continue  # Only executed if the inner loop didn't break
    break  # Stop the outer loop
```
